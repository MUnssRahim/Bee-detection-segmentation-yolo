```python
from ultralytics import YOLO
import mlflow

# Set MLflow tracking path
mlflow_path = "/kaggle/working/bee_models_mlflow"
mlflow.set_tracking_uri(mlflow_path)

# Load your previously trained segmentation model
model = YOLO("/kaggle/working/bee_models/bee_only_seg7/weights/best.pt")

# Train with color-focused augmentation
model.train(
    data="/kaggle/input/bees-segmentation/Segmentation Data/yaml.yaml",
    epochs=35,           # enough to fine-tune
    imgsz=800,          # adjust if OOM occurs
    batch=4,             # reduce if memory error
    lr0=0.0001,
    project="/kaggle/working/bee_models",
    name="bee_only_seg_color_aug",
    augment=True,
    # === Color-based augmentations ===
    hsv_h=0.05,          # hue ±5% (default is 0.015)
    hsv_s=0.8,           # saturation ±80% (default 0.7)
    hsv_v=0.6,           # brightness ±60% (default 0.4)

    # === Other useful augmentations ===
              # small rotation
)
```
