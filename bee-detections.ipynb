```python
import yaml
import os
import mlflow
from ultralytics import YOLO

# 1. SETUP MLFLOW
mlflow_save_path = "/kaggle/working/mlruns"
if not os.path.exists(mlflow_save_path):
    os.makedirs(mlflow_save_path)

mlflow.set_tracking_uri(f"file://{mlflow_save_path}")
mlflow.set_experiment("Drone_Varroa_Detection")

# 2. FIX THE YAML PATHS
input_yaml = "/kaggle/input/drone-varroa-bees/data.yaml"
with open(input_yaml, 'r') as f:
    config = yaml.safe_load(f)

config['train'] = '/kaggle/input/drone-varroa-bees/train/images'
config['val']   = '/kaggle/input/drone-varroa-bees/valid/images'
config['test']  = '/kaggle/input/drone-varroa-bees/test/images'

with open('/kaggle/working/fixed_drone.yaml', 'w') as f:
    yaml.dump(config, f)

# 3. START TRAINING (FIXED MODEL NAME)
# Use 'yolo12s.pt' to load the official pretrained weights
model = YOLO('yolo12s.pt') 

model.train(
    data='/kaggle/working/fixed_drone.yaml',
    epochs=50,
    imgsz=920,
    batch=4,
    lr0=0.001,
    cos_lr=True,
    patience=20,
    augment=True,
    scale=0.5,
    mosaic=1.0,
    mixup=0.1,
    project="/kaggle/working",
    name="drone_bee_Varroa_v3",
    close_mosaic=10
)
```

```python
import shutil
import os

src_file = "/kaggle/working/drone_bee_Varroa_v36/weights/best.pt"
zip_path = "/kaggle/working/best.pt.zip"

# Create zip containing only best.pt
shutil.make_archive(
    zip_path.replace(".zip", ""),
    "zip",
    root_dir=os.path.dirname(src_file),
    base_dir=os.path.basename(src_file)
)

# Verify
print("ZIP created:", os.path.exists(zip_path))
print(os.listdir("/kaggle/working"))
```
